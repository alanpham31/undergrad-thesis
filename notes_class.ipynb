# hypothesis testing
testing to see if something if T/F<br>
null (H0): we can't prove thaht our idea is right. we assume that we are wrong/the wrong is boring/there is no effect, and hope that the data "forces" us to say otherwise<br>
alternative (H1): intersting explaination in which there really is a difference in groups<br>
p-value: conditional probability. specifically, probability of observing your data given that the null hypothesis is true (typically p < 0.05) <br>
alpha: value below which you reject the null. decide before you analyze your data
